```python
from sklearn.model_selection import train_test_split
from sklearn import *
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import *
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
```

```python
# Learn morea about Code Cells: https://quarto.org/docs/reference/cells/cells-jupyter.html

# Include and execute your code here
df = pd.read_csv("https://raw.githubusercontent.com/byuidatascience/data4dwellings/master/data-raw/dwellings_ml/dwellings_ml.csv")
df['before1980'] = df['before1980'].astype('category')
df.head(5)
```

```python
(
ggplot(df, aes(x='before1980', y='livearea')) + \
    geom_boxplot() + \
    ggtitle('Living Area Distribution by Year Built') + \
    xlab('Year Built') + \
    ylab('Living Area (sq ft)') + \
    scale_x_discrete(labels=['After 1980', 'Before 1980'])
)
```

```python
df_before_1980 = df[df['before1980'] == 1]
(
ggplot(df_before_1980, aes(x='numbdrm')) + \
    geom_histogram(binwidth=1, fill='yellow', color='black', alpha=0.8, center=0) + \
    ggtitle('Number of Bedrooms (Built Before 1980)') + \
    xlab('Number of Bedrooms') + \
    ylab('Count')
)
```

```python
df_after_1980 = df[df['before1980'] == 0]
(
ggplot(df_after_1980, aes(x='numbdrm')) + \
    geom_histogram(binwidth=1, fill='blue', color='black', alpha=0.8, center=0) + \
    ggtitle('Number of Bedrooms (Built After 1980)') + \
    xlab('Number of Bedrooms') + \
    ylab('Count')
)
```

```python
X_pred = df.drop(df.filter(regex='before1980|yrbuilt|parcel').columns, axis=1)
y_pred = df['before1980']
X_train, X_test, y_train, y_test = train_test_split(X_pred, y_pred, test_size=0.34, random_state=76)
```

```python
clf = DecisionTreeClassifier(max_depth=12, random_state=76)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print("Decision Tree Classification Report:")
classification_report(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
print(metrics.classification_report(y_pred, y_test))
print(f"Accuracy: {accuracy:.2f}")
```

```python
rf_clf = RandomForestClassifier(n_estimators=100, max_depth=29, random_state=76)
rf_clf.fit(X_train, y_train)
y_pred_rf = rf_clf.predict(X_test)
print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf))
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print(f"Accuracy: {accuracy_rf:.2f}")
```
